```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("../data/ecommerce_sales.db")
```

```python
df = pd.read_csv("../data/ecommerce_sales_analytics_5000.csv")
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df.to_sql(
    "ecommerce_sales",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT
    COUNT(*) AS total_orders,
    SUM(revenue) AS total_revenue,
    AVG(revenue) AS average_order_revenue
FROM ecommerce_sales;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    product_category,
    COUNT(*) AS orders,
    SUM(quantity) AS total_quantity,
    ROUND(SUM(revenue), 2) AS total_revenue,
    ROUND(AVG(revenue), 2) AS average_order_revenue
FROM ecommerce_sales
GROUP BY product_category
ORDER BY total_revenue DESC;
"""

category_sql = pd.read_sql_query(query, conn)

category_sql
```

```python
query = """
SELECT
    region,
    COUNT(*) AS orders,
    ROUND(SUM(revenue), 2) AS total_revenue,
    ROUND(AVG(revenue), 2) AS average_order_revenue
FROM ecommerce_sales
GROUP BY region
ORDER BY total_revenue DESC;
"""

region_sql = pd.read_sql_query(query, conn)

region_sql
```

```python
query = """
SELECT
    customer_id,
    COUNT(*) AS orders,
    ROUND(SUM(revenue), 2) AS total_revenue,
    ROUND(AVG(revenue), 2) AS average_order_value
FROM ecommerce_sales
GROUP BY customer_id
ORDER BY total_revenue DESC
LIMIT 10;
"""

top_customers_sql = pd.read_sql_query(query, conn)

top_customers_sql
```

```python
query = """
SELECT
    strftime('%Y-%m', order_date) AS month,
    COUNT(*) AS orders,
    ROUND(SUM(revenue), 2) AS total_revenue
FROM ecommerce_sales
GROUP BY month
ORDER BY month;
"""

monthly_sql = pd.read_sql_query(query, conn)

monthly_sql.head()
```
